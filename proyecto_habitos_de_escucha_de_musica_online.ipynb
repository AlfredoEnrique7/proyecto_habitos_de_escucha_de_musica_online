#### Examen de la lista duplicados implícitos del género `hiphop`
En esta sección se examina la lista para identificar **duplicados implícitos** del género `hiphop`. Se espera detectar los nombres mal escritos o variantes que hacen referencia al mismo género musical, como los siguientes:
* `hip`  
* `hop`  
* `hip-hop`  

Para solucionarlo:

1. Define una función llamada `replace_wrong_values()` que reciba los siguientes parámetros:
* `df`: el DataFrame a modificar
* `column`: el nombre de la columna a trabajar
* `wrong_values`: una lista con los valores incorrectos
* `correct_value`: el valor correcto para reemplazar

2. Dentro de la función, usa un bucle `for` para iterar sobre cada valor incorrecto y aplicar `.replace()`.
  

**# Definición de una función para reemplazar los duplicados implícitos.**

```python
# 1. Se define una función con sus cuatro parámetros.
def replace_wrong_values(df, column, wrong_values, correct_value):
    
    # 2. Se incluye un bucle for para iterar sobre cada valor incorrecto y aplicar el método .replace()
    for wrong_value in wrong_values:
        df[column] = df[column].replace(wrong_value, correct_value)
        
    # Se retorna el DataFrame con los datos ya modificados
    return df
```

**Eliminación de los duplicados implícitos**
Se invoca la función `replace_wrong_values` pasando:
* `df` como el DataFrame
* `'genre'` como nombre de columna
* `['hip', 'hop', 'hip-hop']` como lista de valores incorrectos
* `'hiphop'` como valor correcto

```python
# Se llama la función 'replace_wrong_values' para unificar las variantes del género hiphop
df = replace_wrong_values(df, 'genre', ['hip', 'hop', 'hip-hop'], 'hiphop')
```

Se cambia el criterio de agrupación a la columna que representa el día de la semana y se filtra solo para los días que te interesan.

```python
# 1. Se filtra el DataFrame para conservar solo lunes y viernes.
df_lunes_viernes = df[df['day'].isin(['Monday', 'Friday'])]

# 2. Se agrupan por el día de la semana y se cuentan las canciones utilizando 'track'.
canciones_por_dia = df_lunes_viernes.groupby('day')['track'].count()

# 3. Se muestran los resultados
print(canciones_por_dia)
```

* En **Shelbyville**, aunque el volumen total es menor, el comportamiento suele ser más estable, y en ocasiones los lunes muestran una actividad competitiva, lo que podría indicar que sus habitantes escuchan música de forma más uniforme durante la jornada laboral o escolar.

#### Combinación de los criterios: día y ciudad.
Para cumplir con esta solicitud se crea una función llamada `number_tracks()` que reciba los dos parámetros:
* `day`: un día de la semana (por ejemplo, `'Monday'`)
* `city`: el nombre de una ciudad (por ejemplo, `'Springfield'`)

Dentro de la función:
1. Se filtra el DataFrame por el día.
2. Luego, filtra por la ciudad.
3. Se cuenta el número de veces que aparece `'user_id'` en ese filtro.
4. Se recupera el número calculado en el paso anterior como resultado.


```python
# Declaración de la función number_tracks() con los dos parámetros: day= y city=.
def number_tracks(day, city):

    # Se almacenan las filas del DataFrame donde el valor en la columna 'day' es igual al parámetro day=.
    filtro_day = df[df['day'] == day]

    # Se filtran las filas donde el valor en la columna 'city' es igual al parámetro city=.
    filtro_city = filtro_day[filtro_day['city'] == city]

    # Se extrae la columna 'user_id' de la tabla filtrada y aplica el método count().
    conteo_tracks = filtro_city['user_id'].count()

    # Se devuelve el número de valores de la columna 'user_id'.
    return conteo_tracks
```

```python
# Se muestra el número de canciones reproducidas en Springfield el lunes.
print("Springfield el lunes:", number_tracks('Monday', 'Springfield'))
```

**Número de canciones reproducidas en Shelbyville el lunes.**

**Número de canciones reproducidas en Shelbyville el Lunes.**

```python
# Se muestra el número de canciones reproducidas en Shelbyville el Viernes.
print("Shelbyville el viernes:", number_tracks('Friday', 'Shelbyville'))
```

* En **Shelbyville**, a pesar de tener un volumen menor de datos, el top de géneros principales sigue exactamente la misma tendencia (*pop*, *dance* y *rock*).

Por lo tanto, **la Hipótesis 2 es rechazada/refutada parcialmente**, ya que los habitantes de ambas ciudades comparten preferencias musicales muy similares al iniciar la semana, en lugar de escuchar géneros distintos.
